### Task 6

```python
#print the 'total_profit' and 'month_number' columns
print(df[['month_number', 'total_profit']])
```

```python
x = df['month_number'].to_numpy() #converting df values into array to be accessed by graph
y = df['total_profit'].to_numpy() #putting into a variable
print(x)
print(y)
```

## Data Visualization (Creating Graphs)

### Task 7

```python
#must include X label name = Month number | Y label name = Total profit | Title = Company profit per month.

plt.bar(x, y) #gathering data points
plt.xlabel('Month Number')
plt.ylabel('Total Profit')
plt.title('Company Profit per Month')
plt.grid(alpha=0.8, linestyle='--')
plt.show()
```
