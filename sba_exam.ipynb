```python
x = df['month_number'].to_numpy() #converting df values into array to be accessed by graph
y = df['total_profit'].to_numpy() #putting into a variable
print(x)
print(y)
```

## *Data Visualization (Creating Graphs)*

### Task 7

```python
#must include X label name = Month number | Y label name = Total profit | Title = Company profit per month.

plt.bar(x, y) #gathering data points
plt.xlabel('Month Number')
plt.ylabel('Total Profit')
plt.title('Company Profit per Month')
plt.grid(alpha=0.8, linestyle='--') #adding grid for easier readability
plt.show()
```

### Task 8
Line Plot must include:
- line style = dotted and line color = red ✅
- Show the legend at the `lower-right` location. ✅
- X label name = Month number | Y label name = Total profit | Title = Company profit per month ✅
- Add a circle marker; the line marker color = black. ✅
- Line width = 3 ✅

```python
plt.figure(figsize=[8,4])
plt.plot(
    x, 
    y, 
    color='red', 
    linestyle='--',
    linewidth=3,
    label='Profit Data of Last Year', #label needed to create legend
    marker='o', #adding circle marker
    mfc='black' #changed color of entire circle (face) to black
    )
plt.legend(loc='lower right') #lower right specification
plt.xlabel('Month Number')
plt.ylabel('Total Profit')
plt.title('Company Profit per Month')
plt.ylim(10000, 50000) #setting range for y-axis
plt.show()
```

### Task 9
Multi-line Plot must include:
- show number of units sold for `each product` (separate line graphs)
- use a legend
- use circular markers

```python
#creating variables for each product and turning data into accessible arrays
# x = month | y = total_profit
#grab columns: 'facecream', 'facewash', 'toothpaste', 'bathingsoap', 'shampoo', 'moisturizer'
fw = df['facewash'].to_numpy() #reassigning from total profits to total units of each product
print(fw)
fc = df['facecream'].to_numpy()
tp = df['toothpaste'].to_numpy()
soap = df['bathingsoap'].to_numpy()
shampoo = df['shampoo'].to_numpy()
moisturizer = df['moisturizer'].to_numpy()
```

```python
plt.plot(x, fc, label='Face Cream Sales Data', marker='o')
plt.plot(x, fw, label='Face Wash Sales Data', marker='o')
plt.plot(x,tp, label='Toothpaste Sales Data', marker='o')
plt.plot(x, soap, label='Bathing Soap Sales Data', marker='o')
plt.plot(x, shampoo, label='Shampoo Sales Data', marker='o')
plt.plot(x, moisturizer, label='Moisturizer Sales Data', marker='o')
plt.legend()
plt.ylim(0, 18000) #setting limits for y- axis range 
plt.xlabel('Month Number')
plt.ylabel('Sales Units in Number')
plt.title('Sales Data')
plt.show()
```
